<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">1. Notions essentielles</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;">
<span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span>
    
<div style="margin-top:9px;">

Une **instruction conditionnelle**, ou instruction de test, permet de *faire des choix* en fonction de la valeur d’une *condition*. On parle souvent d’une instruction *si-alors*, ou *if-else* en anglais.

Une **condition** est une instruction qui est soit vraie, soit fausse. On dit qu’il s’agit d’un *booléen*.</div></div>

**Exemple** : Par exemple, si une variable `a` vaut 5 et qu'une variable `b` vaut 10 alors la condition `a < b` est une condition qui est vraie, tandis que les conditions `a > b` et `a == b` sont fausses.  

On peut tester cela en exécutant les cellules suivantes :

````python
a = 5
b = 10
````

````python
a < b
````

````python
a > b
````

````python
a == b
````

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
<span style="font-size:1.5em">⚠️</span> **Source d'erreur classique** : Le test d'égalité entre deux variables se fait avec un double égal `==` (car le simple `=` a un rôle différent : celui d'affecter une valeur à une variable).</div></div>

Pour tester des inégalité larges (comme $a \leq b$ et $a \geq$ b) ou l'absence d'égalité (comme $a\neq b$) on utilise les syntaxes suivantes :
- le signe `<=` pour *inférieur ou égal*
- le signe `>=` pour *supérieur ou égal*
- le signe `!=` pour *n'est pas égal à*

````python
a <= b
````

````python
a >= b
````

````python
a != b
````

On peut résumer les tests possibles dans le tableau ci-dessous :

| Test | Syntaxe Python |
| --- | --- |
| $$a=b$$ | `a == b` |
| $$a\neq b$$ | `a != b` |
| $$a<b$$ | `a < b` |
| $$a\leq b$$ | `a <= b` |
| $$a>b$$ | `a > b` |
| $$a\geq b$$ | `a >= b` |
| $$a<b<c$$ | `a < b < c` |
| $$a\leq b \leq c$$ | `a <= b <= c` |
| $$a<b\leq c$$ | `a < b <= c` |


## Exemple concret


On souhaite connaître le vainqueur d’un match de football connaissant le score des deux équipes A et B . À la fin de l'*algorithme* suivant , la variable `vainqueur` contient la réponse à la question :

```
si scoreA > scoreB alors
    v ← "équipe A"     (la flèche ← signifie "prend la valeur")
sinon 
    v ← "équipe B"
Afficher v
```

Cet algorithme s'écrirait en Python de la façon suivante :

```python
if scoreA > scoreB:
    v = "équipe A"
else:
    v = "équipe B"
print(v)
```


<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
Dans ce programme, il y a une condition `scoreA > scoreB`. Si elle est vraie, alors le vainqueur sera l'équipe A, sinon ce sera l'équipe B. 

On voit bien qu’un choix est effectué (pour la variable `vainqueur`) selon la valeur de la condition.</div></div>

Notre algorithme n'est pas optimal car il y a un troisième cas possible, celui du match nul. On peut l'adapter de la façon suivante pour en tenir compte :

```
si scoreA > scoreB alors
    v ← "équipe A"     (la flèche ← signifie "prend la valeur")
sinon si scoreA < scoreB alors
    v ← "équipe B"
sinon
    v ← "Match nul"
```

Cet algorithme s'écrirait en Python de la façon suivante :

```python
if scoreA > scoreB:         # si scoreA > scoreB
    v = "équipe A"
elif scoreA < scoreB:       # sinon si scoreA < scoreB
    v = "équipe B"
else:                       # sinon
    v = "Match nul"
```


<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
Dans ce nouveau programme, il y a deux conditions : `scoreA > scoreB` et `scoreA < scoreB`. Si la première est vraie, alors le vainqueur est l'équipe A, sinon on teste si la seconde est vraie et dans l'affirmative le vainqueur est l'équipe B, sinon c'est qu'il y a match nul.</div></div>

<a id="syntaxe"></a>

<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">2. La syntaxe en Python</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">Pour écrire des instructions conditionnelles en Python, il faut utiliser la syntaxe suivante :

```python
if condition1:
	bloc_instructions_1
elif condition2:
	bloc_instructions_2
else:
	bloc_instructions_3
```
    
Les mots-clés `if`, `elif` (contraction de *else if*) et `else` sont les traductions respectives de "si", "sinon si" et "sinon".</div></div>

**Remarques importantes** :

- ⚠️ Il ne faut pas oublier les deux points (`:`) à la fin des lignes avec `if`, `elif` et `else`
- Il est nécessaire de spécifier une condition après les mots clés `if` et `elif`, mais il ne faut pas en donner après le mot clé `else` 
- Les blocs d'instructions à effectuer selon le cas sont **indentés** (“décalage vers la droite”). Ces indentations sont obligatoires dans la syntaxe Python et se font automatiquement si le `:` n’a pas été oublié.
- Les mots clés `elif` et `else` sont optionnels, c'est-à-dire que l’on peut avoir seulement un `if` ou un `if` suivi d'un `else` comme ceci :

```python
if condition:
    bloc_instructions
```

ou 

```python
if condition:
    bloc_instructions_1
else:
    bloc_instructions_2
```

- On peut ajouter autant d'instructions `elif` que l'on souhaite (dépend du nombre de cas à considérer)

Parfois il est nécessaire de combiner plusieurs conditions dans un test.

**Exemple** : à la fac, pour valider son semestre il est parfois nécessaire d’obtenir au moins la moyenne (soit 10/20) dans **toutes** les matières. De manière simplifiée, on suppose ici qu’il n’y a que trois matières (A, B et C). Ainsi, à la fin de l’algorithme suivant, la variable `resultat` vaut `True` si le semestre est validé et `False` sinon.

```
si noteA >= 10 et noteB >= 10 et noteC >= 10 alors
    valide = True
sinon
    valide = False
```

Il est nécessaire que les trois conditions `noteA >= 10`, `noteB >= 10` et `noteC >= 10` soient vraies pour que le trimestre soit validé.

En Python, on utilise le mot clé `and` pour traduire le "et" et le mot clé `or` pour traduire le "ou".

| Condition | Syntaxe Python | Interprétation |
| --- | --- | --- |
| condition1 et condition2 | `condition1 and condition2` | est vraie si les deux conditions sont vraies |
| condition1 ou condition2 | `condition1 or condition2` | est vraie si au moins l'une des deux conditions est vraie |
