---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">
    
Une **machine à sous** est un jeu de hasard. Le joueur appuie sur un bouton et trois symboles s'affichent aléatoirement. Selon la combinaison obtenue, il peut gagner ou perdre des crédits :

- ⭐⭐⭐ → **Super Jackpot !** +20 crédits
- Trois symboles identiques (autre) → **Jackpot !** +10 crédits
- Deux symboles identiques → **Petit gain** +2 crédits
- Trois symboles différents → **Perdu...** −1 crédit

Le joueur commence avec **10 crédits** et joue tant qu'il en a et qu'il le souhaite.</div></div>
