<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">1. Notions essentielles</h3>

---

Dans tout langage de programmation, de manière à organiser le code écrit, on regroupe des fonctionnalités importantes dans ce qu'on appelle des **fonctions**.

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">Une <strong>fonction</strong> est :<br><ul style='margin:6px 0 0 0;padding-left:20px;'><li style='margin:4px 0;'>une manière de <strong>regrouper une séquence d'instructions</strong></li><li style='margin:4px 0;'>de lui donner un <strong>nom</strong></li><li style='margin:4px 0;'>de manière à pouvoir l'<strong>exécuter plus simplement</strong> par la suite, sans avoir à réécrire toutes les instructions à chaque fois.</li></ul></div></div>

## Exemple - l'intérêt d'une fonction



Sur un site de vente en ligne, il est possible de modifier la quantité d'un article (souvent avec des boutons **+** et **−**).

Le site affiche le prix à payer en fonction de la quantité sélectionnée. Voici un programme qui calcule le prix total :

```python
prix_unitaire = 1.5
quantite = 4
prix_total = prix_unitaire * quantite
```

<a id="ex1"></a>
<div style="background-color:#EA8556;color:white;border-radius:4px;padding:10px 18px;margin:20px 0 10px 0;font-family:sans-serif;"><strong style="font-size:1.05em;">Exercice 1 - Révisions</strong></div>

✍️ **Question 1** : Combien de variables sont utilisées dans ce programme ? Comment s'appellent-elles ?

✍️ **Question 2** : Que vaut la variable `prix_total` après exécution du programme ?

**Et si on veut calculer le prix pour d'autres articles ?**

Il faut réécrire ces instructions pour chaque article, en modifiant les valeurs à chaque fois :

````python
# prix d'un autre article
prix_unitaire_2 = 2.5
quantite_2 = 3
prix_total_2 = prix_unitaire_2 * quantite_2
print(prix_total_2)

# puis d'un autre
prix_unitaire_3 = 4.99
quantite_3 = 2
prix_total_3 = prix_unitaire_3 * quantite_3
print(prix_total_3)

# et encore d'un autre :
prix_unitaire_4 = 0.65
quantite_4 = 5
prix_total_4 = prix_unitaire_4 * quantite_4
print(prix_total_4)
````

C'est un peu rébarbatif ! C'est là que les fonctions trouvent leur intérêt : elles permettent d'**écrire les instructions une seule fois pour toutes**, puis de les **exécuter sans avoir à tout réécrire à chaque fois**.

Avant d'écrire une fonction, il faut réfléchir aux informations nécessaires pour effectuer le calcul. Ici, il suffit de connaître le **prix unitaire** et la **quantité** souhaitée. Ce seront les **arguments** de la fonction.

Voici une fonction qui calcule le prix total d'un article :

````python
def prix_a_afficher(prix_unitaire, quantite):
    prix_total = prix_unitaire * quantite
    return prix_total
````

On peut maintenant calculer le prix de chaque article simplement, en **appelant** la fonction :

````python
prix_a_afficher(1.5, 4)
````

````python
prix_a_afficher(2.5, 3)
````

````python
prix_a_afficher(4.99, 2)
````

````python
prix_a_afficher(0.65, 5)
````
