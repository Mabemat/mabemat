- A → D

- X → A

<h4 style="color:#4A5D7A;font-family:sans-serif;font-weight:bold;font-size:1.1em;margin:18px 0 8px 0;padding:7px 14px;background-color:#EBF0F7;border-left:4px solid #4A5D7A;border-radius:0 4px 4px 0;">Le code ASCII</h4>

Chaque caractère (lettre, chiffre, symbole) est représenté dans l'ordinateur par un **nombre** appelé **code ASCII**. Par exemple :
- la lettre `'A'` a le code 65
- la lettre `'B'` a le code 66
- la lettre `'Z'` a le code 90
- la lettre `'a'` a le code 97

Python propose deux fonctions pour travailler avec ces codes :
- `ord(caractere)` : renvoie le code ASCII d'un caractère
- `chr(nombre)` : renvoie le caractère correspondant à un code ASCII

L'astuce : l'opérateur modulo `%`</h4>

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
L'opérateur **modulo** `%` donne le **reste** de la division entière. Par exemple : `7 % 3` vaut 1 car 7 = 2 × 3 + **1**.

En utilisant le modulo 26 (car l'alphabet a 26 lettres), on peut "boucler" sur l'alphabet :
- On ramène d'abord la lettre à une position entre 0 et 25 : `position = ord(lettre) - ord("A")`
- On ajoute le décalage et on prend le modulo 26 : `nouvelle_position = (position + decalage) % 26`
- On reconvertit : `chr(nouvelle_position + ord("A"))`</div></div>
