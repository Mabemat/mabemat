<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">1. Présentation du jeu</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">
    
Le **Pendu** est un jeu classique de devinette de mots. L'ordinateur choisit un mot secret et le masque avec des tirets. Le joueur propose des lettres une par une :

- Si la lettre **est** dans le mot → elle se révèle à toutes ses positions
- Si la lettre **n'est pas** dans le mot → on compte une erreur

Le joueur a **6 erreurs maximum**. S'il trouve le mot avant d'épuiser ses essais, il gagne !

Exemple avec le mot `PYTHON` :
```
_ _ _ _ _ _      → début
_ _ _ H _ _      → après avoir proposé H
P _ _ H _ N      → après avoir proposé P et N
P Y T H O N      → mot trouvé !
```</div></div>

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
Ce projet mobilise :
- l'opérateur **`in`** pour tester si une lettre est dans une chaîne
- **l'indexation** des chaînes (`mot[i]`) pour accéder à chaque lettre
- la **concaténation** de chaînes pour construire le mot masqué lettre par lettre
- les **boucles `while`** et **`for`** pour gérer les tours de jeu
- les **fonctions** pour organiser le code</div></div>

2. Découverte : l'opérateur `in` et l'indexation</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">
    
L'opérateur **`in`** permet de tester si une lettre est présente dans une chaîne de caractères. Il renvoie `True` ou `False` :

```python
"P" in "PYTHON"   # True
"Z" in "PYTHON"   # False
```

On peut aussi accéder à **une lettre précise** d'une chaîne grâce à son **indice** (sa position, à partir de 0) :

```python
mot = "PYTHON"
print(mot[0])   # affiche P
print(mot[1])   # affiche Y
print(mot[5])   # affiche N
```</div></div>
