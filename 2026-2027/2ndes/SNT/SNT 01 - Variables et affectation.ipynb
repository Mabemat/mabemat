<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">3. Variables : types et noms</h3>

---

En informatique, on manipule des informations de **natures très différentes** : un smartphone stocke des numéros, des messages, des photos… Chacune de ces informations est rangée dans une **variable**.

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">Une <strong>variable</strong> est un espace de stockage dans la mémoire de l'ordinateur. 
    
Chaque variable est caractérisée par trois choses : son <strong>nom</strong>, son <strong>type</strong> et sa <strong>valeur</strong>.</div></div>

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">La valeur stockée dans une variable peut <em>changer</em> au cours du programme - c'est pour ça qu'on parle de <em>variable</em>.</div></div>

## Types de variables

Le **type** d'une variable indique la nature de l'information qu'elle contient.

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;"><ul style="margin:0;padding-left:20px;"><li style="margin:4px 0;"><code>int</code> : nombres <strong>entiers</strong> (ex : 5, −12, 0) - <em>integer</em> en anglais</li><li style="margin:4px 0;"><code>float</code> : nombres <strong>décimaux</strong> (ex : 3.14, −0.5) - <em>floating point</em> en anglais</li><li style="margin:4px 0;"><code>str</code> : <strong>chaînes de caractères</strong> (ex : <code>"Bonjour"</code>) - <em>string</em> en anglais</li>
<li style="margin:4px 0;"><code>bool</code> : <strong>booléens</strong>, qui ne peuvent valoir que <code>True</code> (vrai) ou <code>False</code> (faux) - <em>boolean</em> en anglais</li></ul></div></div>

## Noms de variables

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">Jusqu'ici nos variables s'appelaient <code>a</code>, <code>x</code>… Pour des exemples simples, c'est acceptable. Mais si une variable représente quelque chose de concret, donnez-lui un <strong>nom représentatif</strong>.</div></div>

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;"><ol style="margin:0;padding-left:20px;"><li style="margin:4px 0;">Un nom ne peut contenir que des <strong>lettres</strong> (a-z, A-Z), des <strong>chiffres</strong> (0-9) et le caractère <code>_</code>.</li><li style="margin:4px 0;">Un nom <strong>ne peut pas commencer par un chiffre</strong>.</li><li style="margin:4px 0;">Les noms sont <strong>sensibles à la casse</strong> : <code>score</code> et <code>Score</code> sont deux variables différentes.</li><li style="margin:4px 0;"><em>(Bonne pratique)</em> Choisissez un nom <strong>représentatif</strong>.</li><li style="margin:4px 0;"><em>(Bonne pratique)</em> Évitez les <strong>accents</strong> dans les noms.</li></ol></div></div>

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">On utilise <code>_</code> pour séparer plusieurs mots : <code>nb_eleves</code> est bien plus lisible que <code>nbeleves</code>.</div></div>

---

Pour les chaînes de caractères, deux opérations spéciales existent :

- La **concaténation** : mettre deux chaînes bout à bout avec `+`
- La **répétition** : répéter une chaîne plusieurs fois avec `*`

**Exemple** : exécutez le programme suivant.

````python
prenom = "Py"
suffixe = "thon"

langage = prenom + suffixe  # concaténation
print(langage)

rire = 3 * "ha"  # répétition
print(rire)
````

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">La fonction <code>len(chaine)</code> renvoie le <strong>nombre de caractères</strong> d'une chaîne.<br><br><code>len("Bonjour")</code> → <code>7</code><br><code>len("")</code> → <code>0</code> (chaîne vide)<br><br>Elle fonctionne aussi sur d'autres types de collections que vous verrez plus tard.</div></div>

````python
mot = "Python"
print(len(mot))   # affiche 6
print(len("SNT"))  # affiche 3
````
