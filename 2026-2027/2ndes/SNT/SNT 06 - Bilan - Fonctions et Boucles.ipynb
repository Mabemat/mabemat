Elle possède un **nom**, des **paramètres** et un **bloc d'instructions** (indenté).  
Elle renvoie généralement un résultat grâce au mot clé `return`.

```python
def nom_fonction(parametre1, parametre2):
    # bloc d'instructions
    return resultat
```
    
**Boucle `for`** : utilisée quand le **nombre de répétitions est connu** à l'avance.
```python
for i in range(n):    
    # bloc répété n fois
```
    
**Boucle `while`** : utilisée quand le **nombre de répétitions est inconnu**. La boucle se répète **tant que** la condition est vraie.
```python 
while condition:    
    # bloc répété tant que la condition est vraie
```</div></div>
