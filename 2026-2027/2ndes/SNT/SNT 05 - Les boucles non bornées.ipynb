<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">1. Introduction</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">
    
Lors de la séance précédente, vous avez découvert les **boucles bornées** (`for`), utilisées lorsque le **nombre de répétitions est connu à l'avance**.

Mais que se passe-t-il quand on ne sait pas combien de fois on va devoir répéter une instruction ?

C'est là qu'interviennent les **boucles non bornées**, aussi appelées boucles `while` (« tant que » en anglais).</div></div>

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">

On utilise une boucle **`while`** lorsqu'on souhaite répéter des instructions **tant qu'une condition est vraie**, sans connaître à l'avance le nombre de répétitions.</div></div>

## Exemple concret

Un compte bancaire contient **1 000 €**. Chaque année, les intérêts font augmenter ce capital de **3 %**.  
On souhaite savoir **au bout de combien d'années** ce capital dépassera **1 500 €**.

On ne peut pas utiliser une boucle `for` ici, car on **ne connaît pas à l'avance** le nombre d'années nécessaires.  
On répète le calcul **tant que** le capital est inférieur ou égal à 1 500 €.

Sans boucle, on serait obligé d'écrire autant de lignes que nécessaire... ce qui est fastidieux :

````python
capital = 1000
annees = 0

capital = capital * 1.03
annees = annees + 1

capital = capital * 1.03
annees = annees + 1

# ... et ainsi de suite, on ne sait pas combien de fois !

print(annees)
````

Avec une boucle `while`, on écrit ceci de manière bien plus élégante :

````python
capital = 1000
annees = 0

while capital <= 1500:
    capital = capital * 1.03
    annees = annees + 1

print(annees)  # nombre d'années nécessaires
print(capital) # capital atteint
````

<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
Dans cet exemple, la condition est `capital <= 1500`. Tant qu'elle est vraie, le bloc d'instructions est répété. Dès qu'elle devient fausse (c'est-à-dire dès que le capital dépasse 1 500 €), la boucle s'arrête.</div></div>

<a id="2_construction_dune_boucle_non_bornée"></a>
<h3 style="color:#D36740;font-family:sans-serif;font-weight:bold;font-size:1.7em;margin:20px 0 12px 0;padding-bottom:5px;border-bottom:2px solid #D36740;">2. Construction d'une boucle non bornée</h3>

---

<div style="background-color:#EEF1F8;border-left:6px solid #8A9CC1;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#8A9CC1;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Définition</span><div style="margin-top:9px;">
    
La syntaxe d'une boucle `while` en Python est la suivante :

```python
while condition:
    bloc_instructions
```

**Fonctionnement** :
1. La `condition` est évaluée.
2. Si elle est **vraie** (`True`), le `bloc_instructions` est exécuté, puis on retourne à l'étape 1.
3. Si elle est **fausse** (`False`), la boucle s'arrête et le programme continue après.

**Remarques importantes** :
- Ne pas oublier le `:` à la fin de la ligne `while`.
- Le `bloc_instructions` doit être **indenté** (décalé vers la droite avec une tabulation).
- Il faut s'assurer que la condition finira par devenir fausse, sinon la boucle tourne à l'infini !</div></div>

<div style="background-color:#EDF7ED;border-left:6px solid #5A9E6F;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#2E7D4F;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Méthode — Le pattern compteur</span><div style="margin-top:9px;">Dans de nombreux programmes utilisant <code>while</code>, on a besoin de <strong>compter</strong> le nombre de répétitions. On utilise alors le <strong>pattern compteur</strong> :<br><br><pre style='background:#f5f5f5;padding:10px;border-radius:4px;'>compteur = 0          # 1. Initialiser AVANT la boucle
while condition:
    ...               # instructions
    compteur = compteur + 1   # 2. Incrémenter À L'INTÉRIEUR
# À la sortie, compteur contient le nombre de passages dans la boucle</pre>Tu retrouveras ce pattern dans presque tous les exercices de ce notebook !</div></div>

## Comparaison avec la boucle `for`

| | Boucle `for` | Boucle `while` |
|---|---|---|
| Quand l'utiliser ? | Nombre de répétitions **connu** | Nombre de répétitions **inconnu** |
| Mot-clé Python | `for ... in range(...):` | `while condition:` |
| Exemple | Afficher les entiers de 1 à 10 | Attendre qu'un capital dépasse un seuil |


<div style="background-color:#FDF6E3;border-left:6px solid #E6B459;border-radius:0 6px 6px 0;padding:14px 18px;margin:12px 0;font-family:sans-serif;"><span style="color:#9a7010;font-weight:bold;font-size:0.92em;text-transform:uppercase;letter-spacing:0.8px;">Remarque</span><div style="margin-top:9px;">
    
**Boucle infinie** : si la condition d'une boucle `while` ne devient jamais fausse, le programme tourne indéfiniment. En cas de boucle infinie dans Capytale, vous pouvez interrompre l'exécution avec le bouton ⏹ ou en utilisant le menu *Noyau > Interrompre*.</div></div>

## Exemple pas à pas

Voici un programme simple qui affiche les puissances de 2 inférieures à 100 :

````python
p = 1
while p < 100:
    print(p)
    p = p * 2
````

On peut représenter l'algorithme correspondant :

```
p ← 1
Tant que p < 100 :
    Afficher p
    p ← p × 2
Fin Tant que
```
